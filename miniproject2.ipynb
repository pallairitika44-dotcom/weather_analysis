```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

print("libraries imported successfully")
```

```python
data ={
    'city':['delhi','mumbai','chennai','kolkota','banglore'],
    'temperature':[42,35,38,33,28],
    'humidity':[30,85,70,75,60],
    'rainfall':[10,200,150,180,90]
}

df=pd.DataFrame(data)
df.to_csv('weather_data.csv',index=False)
print("csv file created")
print(df)
```

```python
df=pd.read_csv('weather_data.csv')

print("===weather Analysis===")
print()

hottest=df.loc[df['temperature'].idxmax(),'city']
coolest=df.loc[df['temperature'].idxmin(),'city']
most_rain=df.loc[df['rainfall'].idxmax(),'city']
humid=df.loc[df['humidity'].idxmax(),'city']

print(f"hottest city in india:{hottest}({df['temperature'].max()}celcious)")
print(f"coolest city in india:{coolest}({df['temperature'].min()}celcious)")
print(f"highest rainfall in india:{most_rain}({df['rainfall'].max()}mm)")
print(f"highest humidity in india:{humid}({df['humidity'].max()}%)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('weather_data.csv')

plt.figure(figsize=(8,5))
plt.bar(df['city'],df['temperature'],color=['red','orange','yellow','green','blue'])
plt.title('city wise temperature',fontsize=14)
plt.xlabel('City',fontsize=12)
plt.ylabel('temperature (Celsius)',fontsize=12)

for i, val in enumerate(df['temperature']):
    plt.text(i, val + 0.3, str(val)+'°C', ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
#rainfall and humidity line chart

df=pd.read_csv('weather_data.csv')

df.plot(kind='line',x='city',y=['rainfall','humidity'],marker='o',figsize=(8,5),title='rainfall vs humidity')

plt.xlabel('City',fontsize=12)
plt.ylabel('Value',fontsize=12)
plt.tight_layout()
plt.show()
```

```python
df=pd.read_csv('weather_data.csv')

print("="*40)
print("Weather Data")
print("="*40)
print(f"\n hottest :{df.loc[df['temperature'].idxmax(),'city']}({df['temperature'].max()}celsius)")
print(f"coolest :{df.loc[df['temperature'].idxmin(),'city']}({df['temperature'].min()}celsius)")
print(f"highest rainfall :{df.loc[df['rainfall'].idxmax(),'city']}({df['rainfall'].max()}mm)")
print(f"highest humidity:{df.loc[df['humidity'].idxmax(),'city']}({df['humidity'].max()}%)")
print("\n===city ranking by temperature===")
ranking=df[['city','temperature']].sort_values('temperature',ascending=False)
ranking.index=range(1,len(ranking)+1)
print(ranking)
print("="*40)

print("\n===city ranking by rainfall===")
ranking2=df[['city','rainfall']].sort_values('rainfall',ascending=False)
ranking2.index=range(1,len(ranking2)+1)
print(ranking2)
print("="*40)
```
